# Experiment 05 — scoring, selection and cash deployment

Based on [04-opt-score-matrix.ipynb](04-opt-score-matrix.ipynb), with execution and accounting from `hyper-ai-v11.py`.

## Goal

Beat the earlier hand-picked 60% 360-day CAGR / 40% 45-day Sortino formula through better scoring, selection and cash usage. Check whether additional features add forward information, whether clips erase useful distinctions, and whether the apparent improvements survive ties, data-quality checks and concentrated profit attribution.

## Method

Use a bounded, predeclared grid of **18 score calibrations**, with six positions and 90-day inverse-variance sizing fixed. Vary CAGR caps 50/100/200% independently from Sortino caps 3/7/12; retain exact 45-day and near-original 46-day controls; compare the previous blended curve and smooth saturation. Separate ratio-floor and missing-component policies. An activity filter is a distinct treatment.

Compare original and blended scores with and without redistribution after blocked top-ups. Redistribution uses only the selected six vaults, preserves deposit checks, settlement obligations, the 98% budget and 33% concentration/TVL limits, and passes targets through normal trade generation. Test feasible-entrant checking, replacement hurdles and five predeclared tie-order seeds separately.

Feature diagnostics use completed T−1 data and per-date cross-sectional relationships with subsequent 14/30/60-day NAV returns. Include Sharpe, Sortino, CAGR, drawdown and Calmar over nine searchable windows. Future labels cannot cross the report end. The indicator inclusion list, return gate, quarantine rules and recorded guard whitelist define the diagnostic universe; individual portfolio exclusions and deposit availability can further restrict executable entries. Plot score-bin outcomes and per-date correlations, and audit coverage and saturation.

Default charts, performance tables and Markdown summaries use **2026-01-01 → 2026-10-05 (end exclusive, UTC)**. Pre-January observations remain available for indicators. Six predeclared controls/treatments also run over all available history; those results remain in separate artefacts. Quarterly comparisons are temporal stability diagnostics, not independent validation. The dataset is frozen when this notebook first loads it; source and guard fingerprints accompany the snapshot. A new source control is mandatory because the current archive can differ from experiment 04.

## Interpretation

Every backtest begins with \$150,000 and no positions. Source fees, gates, execution bands and settlement rules remain in force. Selection changes are research-only compiled functions; no production module is edited. Missing values and the original risk floor are varied separately. The experiment is exploratory and does not establish a prospective winner; historical universe and availability data remain incomplete. Tie seeds are robustness diagnostics, not optimisation candidates.

## Causal feature diagnostics

Features use completed data at T−1; labels cover the next 14, 30 or 60 daily NAV returns. Both raw forward return and forward realised Sharpe are analysed. Future labels are censored at the report end. Every decision date receives equal weight, reducing domination by dates with larger universes. These are gross NAV associations rather than executable portfolio returns.

The source inclusion, return gate, quarantine and recorded whitelist define this diagnostic universe. Individual state exclusions and deposit availability can further restrict trading. Overlapping labels, carried NAV and current archive membership remain limitations. The diagnostics do not feed back into parameters for this run.

```python
from score_calibration_experiment import ic_summary
with compact_output('Forward feature diagnostics'):
    audit_dir=ARTIFACT_DIR/('features-'+fingerprint[:16])
    audit_dir.mkdir(exist_ok=True)
    labels=['ic','bins','correlations','coverage']
    if all((audit_dir/(label+'.csv')).exists() for label in labels):
        ic,bins,correlations,coverage=[pd.read_csv(audit_dir/(label+'.csv'),parse_dates=['Date']) for label in labels]
    else:
        ic,bins,correlations,coverage=feature_audit(strategy,strategy_universe,indicator_data,panel,parameters,chart_start,chart_end)
        for label,frame in zip(labels,[ic,bins,correlations,coverage]):frame.to_csv(audit_dir/(label+'.csv'),index=False)
    feature_summary=ic_summary(ic)
    feature_summary.to_csv(ARTIFACT_DIR/'2026-forward-feature-summary.csv',index=False)
    for outcome in ['Forward return','Forward Sharpe']:
        view=feature_summary[(feature_summary.Outcome==outcome)&feature_summary.Horizon.eq(30)].sort_values('Mean IC',ascending=False).head(12)
        present_table(view.drop(columns=['Outcome','Horizon']).style.format({'Mean IC':'{:+.3f}','Positive IC fraction':'{:.1%}','Mean paired vaults':'{:.1f}','Lower block band':'{:+.3f}','Upper block band':'{:+.3f}'}).background_gradient(subset=['Mean IC'],cmap='RdBu',vmin=-1,vmax=1),'Highest 30-day associations: '+outcome.lower(),'Equal-date mean cross-sectional Spearman IC. Positive means larger historical feature values rank stronger future outcomes. Exploratory 200-draw moving-block bands preserve some overlapping-label dependence; the table is descriptive, not a feature selection rule.',CHART_PERIOD)
        chart_intro('Forward associations: '+outcome.lower(),'Heatmap cells show equal-date mean cross-sectional IC, on a shared −1 to +1 scale. Missing cells lack enough variable, paired observations. Drawdown is stored as a negative return: a larger value means a smaller historical loss.',CHART_PERIOD)
        show_static(ic_figure(feature_summary,outcome),CHART_PERIOD)
    chart_intro('Feature redundancy including Sortino','Average per-date Spearman correlations among eligible candidates. The separate 45-, 90- and 360-day panels show whether alternative metrics duplicate the original inputs. Pairwise missing-value deletion is used; correlations are not predictive IC.',CHART_PERIOD)
    show_static(correlation_figure(correlations),CHART_PERIOD)
    chart_intro('Monotonic feature bins','Average next-30-day NAV returns for within-date feature-rank quintiles. Each date contributes equally within its populated bins; ties can leave bins empty. The highest quintile should improve consistently before a feature is trusted as a ranker.',CHART_PERIOD)
    show_static(quintile_figure(bins),CHART_PERIOD)
    cover=coverage.groupby(['Period','Metric'])[['Eligible vaults','Finite feature']].sum()
    cover['Finite coverage']=cover['Finite feature']/cover['Eligible vaults']
    cover=cover.reset_index()
    present_table(cover.style.format({'Finite coverage':'{:.1%}'}),'Feature coverage among eligible candidates','Denominator is the gate-filtered diagnostic universe on 2026 decision dates, rather than all vault/calendar observations. Undefined ratios and missing history remain separate from zero scores. Young vaults retain the treatment-specific missing policy.',CHART_PERIOD,collapsed=True)
```
