# Experiment 06 — three and four scoring inputs

Based on [04-opt-score-matrix.ipynb](04-opt-score-matrix.ipynb) and [05-opt-score-calibration-selection-cash.ipynb](05-opt-score-calibration-selection-cash.ipynb).

## User goal

Test scoring systems with **three and four distinct inputs**, rather than inferring from the two-input winners that more complex scores cannot improve the strategy. An input is one metric/lookback cell, not a clipping term or a position weight.

## Method

Use the fixed 7 October publication and unchanged v11 universe, selection, guards, costs, two-day cadence, six positions and 90-day inverse-variance sizing. Keep the original unscored/missing behaviour and the parent capacity redistribution. Compare original 360-day CAGR/45-day Sortino scoring and the 360/46-day clipping blend.

Predeclare 18 supports per cardinality and clipping family: **36 three-input and 36 four-input configurations**, plus two two-input controls. For three inputs, allocate 48% to 360-day CAGR, 32% to core Sortino and 20% to one additional feature. For four inputs, keep the same core and allocate 10% to each of two additional features. This holds the total added mass at 20%, preventing negligible extra weights from masquerading as meaningful complexity. Additional cells span CAGR, Sharpe, Sortino, maximum drawdown and Calmar across short and long windows.

Run every configuration on both **2026-01-01 → 2026-10-05** and **2023-03-01 → 2026-10-05**, end exclusive, UTC, with fresh $150,000 starts. All feature observations precede decisions. Additional unavailable metrics propagate an unscored composite; young vaults are not automatically excluded. These periods overlap, so neither is an independent holdout. The initial grid does not optimise coefficients or exhaust all possible supports. Report full-history-selected candidates separately from descriptive 2026 maxima.

## Presentation

Default charts and tables use the 2026 period. Full-history evidence is collapsed and retained in ignored artefacts. Every displayed table and chart has a subheading, description and explicit period. Source/exact-formula controls must agree on full equity and trade ledgers. Examine best-day jumps, leading vaults, curator annotations and BTC/ETH intervals before interpreting a high CAGR as broadly supported.

## Key new insights

- **3 inputs, highest 2026 CAGR:** Blend 46d / 3 inputs / 6m CAGR: 51.99% CAGR, 2.29 Sharpe and -7.29% drawdown. Highest Sharpe is 2.29 for Blend 46d / 3 inputs / 6m CAGR.
- **4 inputs, highest 2026 CAGR:** Legacy 45d / 4 inputs / 6m Sortino + 2m CAGR: 52.30% CAGR, 2.23 Sharpe and -7.62% drawdown. Highest Sharpe is 2.30 for Legacy 45d / 4 inputs / 6m CAGR + 1m Sortino.
- **Joint improvements against original v11:** 1 of 72 added-input configurations have higher 2026 CAGR and Sharpe with no worse drawdown. This does not establish unseen-time improvement.

- **Strongest balanced four-input candidate:** 48% 360-day CAGR + 32% 45-day Sortino + 10% 180-day CAGR + 10% 30-day Sortino, with legacy score clips (CAGR 100%, Sortino 3). It improves 2026 CAGR and Sharpe with slightly better drawdown than original v11; full-history drawdown is slightly worse (-8.08% versus -7.92%). These are four metric/window inputs but only two different metrics.
- **Extra windows matter in this replay:** collapsing the four-input leader to the same 58/42 metric balance on just 360-day CAGR/45-day Sortino reduces 2026 CAGR from 51.77% to 41.89% and Sharpe from 2.30 to 1.99. Full-history CAGR falls from 22.05% to 15.85% and Sharpe from 1.62 to 1.24. This diagnostic supports an incremental benefit from the additional windows, subject to search selection bias.

## Robustness of results

All 74 configurations completed both periods. Three-input scores allocate 20% to one additional metric/window; four-input scores allocate 10% to each of two additional cells. Both keep 80% on the same 60/40 CAGR/Sortino core. This tests a predeclared add-on grid, not all combinations or optimal coefficients. Missing additional metrics propagate an unscored composite, as in the source; sparse long-window coverage can therefore change ranking materially. Full-history selections and 2026 maxima are labelled separately, with no claim of independent validation.

Source v11 and its exact two-input matrix match equity and complete trade ledgers in both periods. Position attribution reconciles to initial capital with a maximum residual of 5.82e-11 USD. The notebook reports best days, excess kurtosis, leading-vault concentration, curator annotations and matching BTC/ETH intervals. Fixed-ledger contributor removals are sensitivities, not executable exclusion backtests.

## Source replication

Check that the two-input wrapper expresses the original strategy before adding inputs.

```python
control_runs = {}
replication = []
def trade_ledger(run):
    return [(t.pair.pool_address,t.opened_at,t.executed_at,t.executed_quantity,t.executed_reserve,t.executed_price) for t in run['result'].state.portfolio.get_all_trades()]
with compact_output('Source and two-input controls'):
    for period in PERIODS:
        row = lookup['Legacy 45d']
        source = evaluator.run(row['Weights'], *PERIODS[period], n=6, sizing='inverse_variance', missing=True, profile='legacy', source_control=True, keep=True)
        exact = evaluate(evaluator, row, period, keep=True)
        assert np.allclose(source['equity'], exact['equity'], rtol=0, atol=1e-6)
        assert trade_ledger(source) == trade_ledger(exact)
        control_runs[period] = source
        replication.append({'Period':period,'Final equity (USD)':source['metrics']['Final equity (USD)'],'Trades':source['metrics']['Trades'],'Full equity and ledger match':True})
    saved = json.loads((RUN_ROOT/'01/metrics.json').read_text())
    for period, old_name in [('2026','Short (source reference)'),('Long','Long (all available vault history)')]:
        for metric in ('Final equity (USD)','CAGR','Sharpe','Maximum drawdown','Trades'):
            assert np.isclose(control_runs[period]['metrics'][metric], saved[old_name][metric], rtol=1e-12, atol=1e-8)
present_table(pd.DataFrame(replication).query("Period == '2026'"), 'Original v11 replication', 'The exact two-input wrapper matches every source trade and the complete equity path. The same checks pass over full history.', SHORT_PERIOD)
(pd.DataFrame(replication)).to_csv(ARTIFACT_DIR/'replication.csv',index=False)
```

## Interpretation and next research step

### Three and four inputs — 2026-01-01 → 2026-10-05 (end exclusive, UTC)

Experiment 04 already evaluated three- and four-input proposals, but only six of each per objective: 12 three-input and 12 four-input proposals in total. It also tested a dense 45-input equal score. Sizing and missing-score semantics varied during that search, and coefficient refinements stayed on the winning support. The two-input selected formulas were therefore not evidence that additional inputs could not help.

Experiment 06 holds sizing and selection fixed. The strongest balanced candidate is:
