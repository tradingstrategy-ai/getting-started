Saved calculations from the same pinned archive; this appendix runs no new study or backtest.2026is shown first, with longer-period results explicitly labelled. Every table has a separate heading and description.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown
ROOT = Path.cwd()
if ROOT.name == 'appendices':
    ROOT = ROOT.parent
elif not (ROOT / 'portfolio_snapshot_data.py').exists():
    ROOT = ROOT / 'scratchpad/hyperliquid-vov-2nd'
ARTIFACT = ROOT / '_artifacts-restored-2026-10-07/10'
period_names = {'2026':'2026-01-01 → 2026-10-05', 'Long':'2023-03-01 → 2026-10-05', 'Pre-2026':'2023-03-01 → 2026-01-01'}
def element(frame, heading, description):
    display(Markdown('### ' + heading + '\n\n' + description))
    if frame.empty:
        display(Markdown('No supported rows.'))
    else:
        display(frame.style.format(precision=5, na_rep='unsupported').hide(axis='index'))

frame = pd.read_parquet(ARTIFACT / 'summaries.parquet')
for period in ('2026','Long','Pre-2026'):
    subset=frame[frame.period.eq(period)]
    for (h,score),group in subset.groupby(['h','score'],sort=True):
        element(group[['policy_id','branch','variant','N','L','estimator','p','mean_return','mean_variance','weighting_lift','selection_lift','mean_cash','resolved_origins','origins','unresolved_weight']], f'{score}; {period_names[period]} UTC (end exclusive); {h}-day horizon', "All registered policy arms, including bounded companions. Returns are simple horizon returns in fractions, not strategy CAGR. Each policy's own-support mean is descriptive; comparison decisions require the paired contrasts.")
```
