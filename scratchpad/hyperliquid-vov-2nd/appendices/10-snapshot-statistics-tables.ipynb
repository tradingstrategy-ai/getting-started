Every table has a separate heading and description.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown
ROOT = Path.cwd()
if ROOT.name == 'appendices':
    ROOT = ROOT.parent
elif not (ROOT / 'portfolio_snapshot_data.py').exists():
    ROOT = ROOT / 'scratchpad/hyperliquid-vov-2nd'
ARTIFACT = ROOT / '_artifacts-restored-2026-10-07/10'
period_names = {'2026':'2026-01-01 → 2026-10-05', 'Long':'2023-03-01 → 2026-10-05', 'Pre-2026':'2023-03-01 → 2026-01-01'}
def element(frame, heading, description):
    display(Markdown('### ' + heading + '\n\n' + description))
    if frame.empty:
        display(Markdown('No supported rows.'))
    else:
        display(frame.style.format(precision=5, na_rep='unsupported').hide(axis='index'))

frame = pd.read_parquet(ARTIFACT / 'intervals.parquet')
for period in ('2026','Long','Pre-2026'):
    group=frame[frame.period.eq(period)]
    element(group, f'{period_names[period]} UTC (end exclusive);30-day primary horizon', 'The18 frozen contrasts use matched origins and shared moving calendar blocks. A descriptive status means the ten-complete-block gate was not met; long/pre-2026 bands remain exploratory.')
for filename, title, description in [
    ('assignment_pooled','Persistent assignment null','Same ex-ante basket and weight multiset; coherent seeded identity priorities remain fixed across dates. Return percentiles and variance percentiles answer different questions.'),
    ('score_null','Conditional score permutation null','These date-wise exchangeability and age/quality-stratified controls are descriptive; they are not persistent policy inference.'),
    ('phases','Origin phase sensitivity','Four overlapping four-day grids are not independent replications.'),
    ('spaced','Non-overlapping-origin sensitivity','Diagnostic thinned origins change the available cohort; compare paired effects and support.'),
]:
    values=pd.read_parquet(ARTIFACT / (filename+'.parquet'))
    if 'period' in values:
        for period in ('2026','Long','Pre-2026'):
            element(values[values.period.eq(period)], title+';'+period_names[period]+' UTC;30-day horizon', description)
    else:
        element(values, title+';2026-01-01 →2026-10-05 UTC;30-day horizon', description)
```
