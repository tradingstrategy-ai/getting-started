Based on [06-opt-score-3-4-inputs.ipynb](06-opt-score-3-4-inputs.ipynb).

## User goal

Repeat the same scoring experiment with **7, 8 and 10 positions allocated**, including the highlighted four-input candidate (360-day CAGR, 45-day Sortino, 180-day CAGR and 30-day Sortino). Keep six positions as the matched reference. An input is a metric/lookback cell, not a distinct metric or a position weight.

## Method

Freeze all 74 experiment 06 supports: two two-input controls, 36 three-input configurations and 36 four-input configurations. Preserve 18 supports per cardinality/family, 60/40 two-input controls and fixed 48/32/20 or 48/32/10/10 coefficients, original and blended clipping recipes, source missing-score behaviour, 90-day inverse-variance sizing, 33% concentration cap, 98% allocation target, parent capacity redistribution and two-day cadence.

Change the maximum positions, maximum capacity positions and entry rank to N; keep the source two-rank exit buffer as N+2. The capacity extension remains disabled because both position limits equal N. Other selector rules, admission guards, universe, thresholds and source-modelled costs remain unchanged. A position limit is a maximum; admission and sizing can produce fewer funded vaults.

Run all supports at N=6, 7, 8 and 10 in **2026-01-01 → 2026-10-05** and **2023-03-01 → 2026-10-05**, end exclusive, UTC, with fresh \$150,000 starts and causal feature observations. There are 296 configurations per period, including 74 verified six-position references, and 444 new replays. Use the exact pinned repaired-data generation from experiment 06. Three independent processes handle the new breadth arms serially, with observable checkpoint progress and separate caches.

## Replication and matching controls

Six-position source checks establish the same starting point as experiment 06.

```python
def trade_ledger(run):
    return [(t.pair.pool_address,t.opened_at,t.executed_at,t.executed_quantity,t.executed_reserve,t.executed_price) for t in run['result'].state.portfolio.get_all_trades()]
replication=[]
with compact_output('Original source and parent replication'):
    row=lookup['Legacy 45d']
    for period in PERIODS:
        source=evaluator.run(row['Weights'],*PERIODS[period],n=6,sizing='inverse_variance',missing=True,profile='legacy',source_control=True,keep=True)
        exact=evaluate(evaluator,row,period,6,keep=True)
        assert np.allclose(source['equity'],exact['equity'],rtol=0,atol=1e-6)
        assert trade_ledger(source)==trade_ledger(exact)
        replication.append({'Period':period,'Final equity (USD)':source['metrics']['Final equity (USD)'],'Trades':source['metrics']['Trades'],'Equity and complete ledger match':True})
present_table(pd.DataFrame(replication).query("Period=='2026'"), 'Original v11 replication', 'Source v11 and the exact two-input wrapper match every trade and the full equity path. The same checks pass over full history.', SHORT_PERIOD)
pd.DataFrame(replication).to_csv(ARTIFACT_DIR/'replication.csv',index=False)
```
