remaining gap is the portfolio wrapper the screen omits: inverse-vol sizing, the momentum

backtest; it does not promise a portfolio edge.

## What is worth keeping

1. **The decision-aligned screen itself** - re-usable, cheap, and demonstrably better than

## Recommendation

Stop iterating on cross-sectional selection signals for this universe/period: two disciplined
passes over ~35 features found none that convert to a robust top-6 edge over the shipped
composite. The evidence points at a ceiling set by the data (one 11-month bear regime, no
out-of-sample) rather than a missing feature. The highest-value next moves are **not** more
selection features but: (a) walk-forward / shadow-live validation of NB36 as new data
accrues; (b) capacity and cost realism (the real live constraint); (c) if selection is
revisited, a graded immature-vault score, screened by precision-at-6 first.
